```python
# ==========================================================
# Project      : Quant Research Lab
# Notebook     : 01_Authentication.ipynb
# Purpose      : Verify Python environment and required packages
# ==========================================================

import sys
import platform

print("="*60)
print("Quant Research Lab - Environment Check")
print("="*60)

print(f"Python Version : {sys.version.split()[0]}")
print(f"Python Path    : {sys.executable}")
print(f"Operating Sys  : {platform.system()} {platform.release()}")

print("\nChecking Required Packages...\n")

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "matplotlib": "matplotlib",
    "scikit-learn": "sklearn",
    "kiteconnect": "kiteconnect",
    "pandas-ta": "pandas_ta"
}

for display_name, import_name in packages.items():
    try:
        module = __import__(import_name)
        version = getattr(module, "__version__", "Version Not Available")
        print(f"✅ {display_name:<15} Installed (Version: {version})")
    except ImportError:
        print(f"❌ {display_name:<15} NOT Installed")

print("\nEnvironment Check Completed.")
print("="*60)
```

```python
import sys
sys.path.append("..")

from src.supertrend import load_raw_data, apply_supertrend, save_processed_data

df = load_raw_data("BHARTIARTL")
df = apply_supertrend(df)
df.tail(15)
```

```python
import matplotlib.pyplot as plt

plot_df = df.tail(150)

fig, ax = plt.subplots(figsize=(16, 7))

ax.plot(plot_df.index, plot_df["close"], label="Close Price", color="black", linewidth=1)
ax.plot(plot_df.index, plot_df["SUPERTl_10_3"], label="Supertrend (Uptrend)", color="green", linewidth=1.5)
ax.plot(plot_df.index, plot_df["SUPERTs_10_3"], label="Supertrend (Downtrend)", color="red", linewidth=1.5)

ax.set_title("BHARTIARTL — 4H Candles with Supertrend (10, 3)")
ax.set_xlabel("Date")
ax.set_ylabel("Price")
ax.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
